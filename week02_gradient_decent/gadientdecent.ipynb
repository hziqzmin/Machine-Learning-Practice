```
import numpy as np
import torch

# 데이터 읽기 함수
def load_dataset(file):

  data = np.loadtxt(file)
  print("DATA=",data)

  x = data[:,0:-1]
  print("X=",x)

  y = np.reshape(data[:,-1],(3,1))
  print("Y=",y)

  # 텐서로 변경
  # 이곳을 채우세요.
  x = torch.tensor(x, dtype=torch.float)
  y = torch.tensor(y, dtype=torch.float)

  return (x, y)

# 텐서를 리스트로 바꾸는 함수
def tensor2list(input_tensor):
  return input_tensor.cpu().detach().numpy().tolist()

x, y = load_dataset("/gdrive/My Drive/colab/week2/train.txt")

# 가중치 초기화
w = torch.randn(2)*3.0
print("\nInit: w = {0}\n".format(tensor2list(w)))

# 학습률
alpha = 0.1

# 학습데이터 수
m = y.size(0)

for epoch in range(50):
  cost = 0
  for i in range(m):
  # 이곳을 채우세요.
    #cost = cost +((w*x[i])-y[i]) * x[i]
    prediction = torch.dot(w, x[i])
    cost = cost + (prediction - y[i]) * x[i]
  # 이곳을 채우세요.
  cost = 1/m*cost

  w = w-alpha * cost

  print("Epoch {0:d}: w = {1}".format(epoch+1,tensor2list(w)))
```
